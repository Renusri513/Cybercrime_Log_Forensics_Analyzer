# Cybercrime Log Forensics Analyzer

This notebook analyzes a local authentication log, builds a timeline, flags repeated failed logins and unusual-hour events, and creates CSV reports and charts.

**Forensic caution:** use only logs you are authorized to examine. Findings are investigative leads, not proof of malicious activity. The included log is synthetic.

## 2. Configure the input and analysis settings

```python
from pathlib import Path
from analyzer import analyze_log

INPUT_FILE = Path('sample_auth.log')  # Change to an authorized log file
OUTPUT_DIR = Path('results')
FAILED_LOGIN_THRESHOLD = 5
UNUSUAL_HOUR_START = 0  # inclusive
UNUSUAL_HOUR_END = 5    # exclusive
SYSLOG_YEAR = 2026      # used for timestamps like 'Oct 22 01:02:11'
```

## 3. Run the analysis

```python
result = analyze_log(
    INPUT_FILE,
    OUTPUT_DIR,
    failed_threshold=FAILED_LOGIN_THRESHOLD,
    unusual_start=UNUSUAL_HOUR_START,
    unusual_end=UNUSUAL_HOUR_END,
    default_year=SYSLOG_YEAR,
)
result
```

## 4. Review the summary and flagged source IPs

```python
print((OUTPUT_DIR / 'summary.txt').read_text(encoding='utf-8'))
import pandas as pd
suspicious_ips = pd.read_csv(OUTPUT_DIR / 'suspicious_ips.csv')
flagged_events = pd.read_csv(OUTPUT_DIR / 'flagged_events.csv')
display(suspicious_ips)
display(flagged_events.head(20))
```
